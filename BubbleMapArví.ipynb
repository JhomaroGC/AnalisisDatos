```python
import pandas as pd
import plotly.express as px
```

```python
# Ejemplo de la estructura de la lista de objetos:
registros_parque = [
    {'Atracción': 'silleta 3', 'Latitud': 6.2654855, 'Longitud': -75.4914161, 'Hora': 7, 'Usuarios': 99993},
    {'Atracción': 'silleta 3', 'Latitud': 6.2654, 'Longitud': -75.4914, 'Hora': 8, 'Usuarios': 120000},
                        ]
```

```python


df = pd.DataFrame(registros_parque)
display(df.head())
```

```python


fig = px.scatter_mapbox(
  df,
  lat="Latitud",
  lon="Longitud",
  size="Usuarios",
  color= "Atracción",
  hover_name="Atracción",
  labels={"Usuarios": "Usuarios"},
  text="Usuarios", # Muestra el número
  animation_frame="Hora", # AQUÍ se crea el slider 6am-9pm
  range_color=[0,10], #[0, df['Usuarios'].max()], # Escala de color fija
  color_continuous_scale="Viridis", # O "Hot" para mapa de calor
  zoom=15,
  mapbox_style="carto-positron"
  )

fig.update_traces(marker=dict(opacity=0.7)) # Círculos semitransparentes
fig.show()
```
